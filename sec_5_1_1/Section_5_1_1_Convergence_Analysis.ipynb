The uniform proposal is used with $w=0.001$. We consider $I\in\{1{,}000,5{,}000,10{,}000,50{,}000,100{,}000\}$ and report pointwise medians and 25--75% interquartile bands over 20 independent replications.


```python
# Loads data/scenario_size.csv; runs the experiment only if that file is absent.
scenario_size_results = run_scenario_size_experiment()
```

```python
scenario_size_figures = plot_scenario_size_experiment(
    scenario_size_results,
    max_iteration=2500,
    title=None,
    axis_label_size=20,
    tick_label_size=20,
    legend_font_size=20,
    show=True,
)
```
