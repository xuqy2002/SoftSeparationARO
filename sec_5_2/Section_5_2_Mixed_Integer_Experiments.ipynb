## Common settings

The paper uses the uniform proposal by default. Set `PROPOSAL_MODE` to `"dual_vmf"` to use the adaptive proposal; `DUAL_VMF_KAPPA=None` then selects $d/2$. By default, Algorithm 3 uses a support-demand LP initialization followed by 100 continuous warm-start iterations with $\alpha_k=1/\sqrt{k}$ and one MH transition per iteration. Both the use of this warm start and its iteration count can be changed below. The reported Algorithm 3 time includes warm start and single-tree certification; the exact robust evaluation and C&CG reference are post-hoc benchmarks.


```python
USE_WARM_START = True
WARMUP_ITERATIONS = 100

exp.W = 1
exp.PROPOSAL_MODE = "uniform"
exp.DUAL_VMF_KAPPA = None
exp.USE_WARM_START = USE_WARM_START
exp.WARMUP_ITERATIONS = WARMUP_ITERATIONS
exp.EPSILON_OPT = 200.0
exp.EPSILON_CERT = 100.0
exp.CERTIFICATION_RHO = 0.9
exp.EPSILON_FALSE = 0.1
exp.MAX_CERTIFICATION_ATTEMPTS = 5_000
exp.MAX_TOTAL_ORACLE_CALLS = 100_000
exp.REUSE = True
```

## Experiment 1: Continuous ellipsoidal uncertainty


```python
ELLIPSOID_METHODS = ("ours", "ldr")
ELLIPSOID_DIMS = (3, 10, 30)

ellipsoid_raw, ellipsoid_summary = exp.run_ellipsoid_experiment(
    methods=ELLIPSOID_METHODS,
    dims=ELLIPSOID_DIMS,
    reps=range(1, 6),
    reuse=exp.REUSE,
)
display(ellipsoid_summary)
```

## Experiment 2: Finite scenario uncertainty


```python
SCENARIO_METHODS = ("ours", "ldr")
SCENARIO_DIMS = (3, 10, 30)
SCENARIO_SIZES = (1_000, 5_000, 10_000, 50_000, 100_000)

scenario_raw, scenario_summary = exp.run_scenario_experiment(
    methods=SCENARIO_METHODS,
    dims=SCENARIO_DIMS,
    scenario_sizes=SCENARIO_SIZES,
    reps=range(1, 6),
    reuse=exp.REUSE,
)
display(scenario_summary)
```

## Experiment 3: Budget uncertainty


```python
BUDGET_METHODS = ("ours", "ldr")
BUDGET_DIMS = (10, 30, 50, 70, 100)
GAMMAS = (0.1, 0.3, 0.5, 0.7, 0.9)

budget_raw, budget_summary = exp.run_budget_experiment(
    methods=BUDGET_METHODS,
    dims=BUDGET_DIMS,
    gammas=GAMMAS,
    reps=range(1, 6),
    reuse=exp.REUSE,
)
display(budget_summary)
```
